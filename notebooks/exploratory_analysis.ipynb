# Exploratory Data Analysis — Resume-Job Fit Dataset

**Project:** AI-Based Resume Screening and Job Fit Prediction

This notebook inspects the dataset before modeling.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_preprocessing import prepare_dataset, parse_skill_list
from src.config import DATASET_PATH

sns.set_style('whitegrid')
print(f'Dataset path: {DATASET_PATH}')
```

```python
df, report = prepare_dataset()
print(f'Shape: {df.shape}')
df.head()
```

```python
df.info()
print('\nMissing values:')
print(df.isnull().sum())
```

```python
score_cols = ['ai_match_score', 'skill_string_match_score', 'fuzzy_match_score']
df[score_cols].describe()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, score_cols):
    df[col].hist(bins=30, ax=ax, color='steelblue', edgecolor='white')
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

```python
print(f'Categories: {df["category"].nunique()}')
df['category'].value_counts().plot(kind='barh', figsize=(10, 8))
plt.title('Job Category Distribution')
plt.show()
```

```python
df['fit_label'].value_counts().plot(kind='bar', color=['#e74c3c', '#f39c12', '#27ae60'])
plt.title('Fit Label Distribution (derived from ai_match_score)')
plt.xlabel('Fit Label')
plt.ylabel('Count')
plt.show()
```

```python
all_skills = []
for val in df['resume_skill_list'].dropna():
    all_skills.extend(parse_skill_list(val))
skill_counts = pd.Series(all_skills).value_counts().head(20)
skill_counts.plot(kind='barh', figsize=(10, 6))
plt.title('Top 20 Resume Skills (from dataset)')
plt.show()
```
